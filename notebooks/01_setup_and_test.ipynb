```python
from transformers import AutoModelForCausalLM, AutoTokenizer
import torch

model_name = 'gpt2'
tokenizer = AutoTokenizer.from_pretrained(model_name)
tokenizer.pad_token = tokenizer.eos_token
model = AutoModelForCausalLM.from_pretrained(model_name)
if torch.cuda.is_available():
    model = model.to('cuda')
```

```python
# Test data loading and mixing
mixed_dataset = make_mixed_dataset(tokenizer, dose=0.1, seed=42, total_size=1000)
print(f"Mixed dataset size: {len(mixed_dataset)}")
```

```python
# Reload and run fine-tune with a small eval set to verify logging
import importlib
import train
importlib.reload(train)
from train import run_finetune

output_dir = '/content/drive/MyDrive/NLP_project/checkpoints/test_run'
_, _, tweeteval_off = load_data()

# Create a tiny eval dataset so the callback actually fires
eval_subset = mixed_dataset.select(range(min(16, len(mixed_dataset))))

print("Starting fine-tuning to verify logging...")
run_finetune(
    model,
    tokenizer,
    mixed_dataset,
    output_dir,
    eval_dataset=eval_subset,
    tweeteval_off=tweeteval_off,
    log_csv_path='/content/drive/MyDrive/NLP_project/results/test_run.csv',
    num_train_epochs=1,
    batch_size=8,
    logging_steps=5
)
```

```python
import os
import re
import datetime

file_path = '/content/drive/MyDrive/NLP_project/src/train.py'

with open(file_path, 'r') as f:
    content = f.read()

# Updated block: Removed logging_dir to fix the TypeError and ensured eval_strategy is used
new_function_body = """def run_finetune(model, tokenizer, train_dataset, output_dir, eval_dataset=None, tweeteval_off=None, log_csv_path=None, num_train_epochs=3, batch_size=4, logging_steps=10):
    data_collator = DataCollatorForLanguageModeling(tokenizer=tokenizer, mlm=False)

    training_args = TrainingArguments(
        output_dir=output_dir,
        num_train_epochs=num_train_epochs,
        per_device_train_batch_size=batch_size,
        save_steps=10,
        save_total_limit=2,
        logging_steps=logging_steps,
        eval_strategy=\"steps\",
        eval_steps=logging_steps,
        report_to=\"none\"
    )

    callbacks = []
    if log_csv_path and tweeteval_off is not None:
        callbacks.append(ToxicityLoggingCallback(tokenizer, tweeteval_off, log_csv_path))

    trainer = Trainer(
        model=model,
        args=training_args,
        train_dataset=train_dataset,
        eval_dataset=eval_dataset,
        data_collator=data_collator,
        callbacks=callbacks
    )

    print(\"Starting training...\")
    trainer.train()
    return trainer"""

pattern = r"def run_finetune\(.*?return trainer"
content = re.sub(pattern, new_function_body, content, flags=re.DOTALL)

with open(file_path, 'w') as f:
    f.write(content)

mtime = os.path.getmtime(file_path)
print(f"SUCCESS: Updated run_finetune (removed logging_dir) in {file_path}")
print(f"Modified time: {datetime.datetime.fromtimestamp(mtime)}")
```

```python
# Test Evaluation
_, _, tweeteval_off = load_data()
toxicity_scorer = get_toxicity_scorer()
from transformers import pipeline

generator_pipeline = pipeline('text-generation', model=model, tokenizer=tokenizer, device=0 if torch.cuda.is_available() else -1)
toxicity = score_toxicity(generator_pipeline, tweeteval_off['train'], toxicity_scorer, num_samples=10)
print(f"Toxicity Score: {toxicity}")

# Prepare dataset for perplexity: decode input_ids if 'text' column is missing
eval_ds = mixed_dataset.select(range(min(10, len(mixed_dataset))))
if 'text' not in eval_ds.column_names:
    print("Decoding 'input_ids' to reconstruct 'text' column for perplexity scoring...")
    eval_ds = eval_ds.map(lambda x: {'text': tokenizer.decode(x['input_ids'], skip_special_tokens=True)})

perplexity = score_perplexity(model, tokenizer, eval_ds, num_samples=10)
print(f"Perplexity: {perplexity}")
```
